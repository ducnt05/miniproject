# E-ELAN Object Detection

Pipeline: E-ELAN Backbone + FPN Neck + single-scale YOLO-style Detection Head.

Class mapping: `0=cat`, `1=dog`, `2=panda`.

```python
from pathlib import Path
import json
import sys
import torch
from torch.utils.data import DataLoader, Subset

PROJECT_ROOT = Path(r"D:/SPCV/miniproject")
DATA_ROOT = PROJECT_ROOT / "datasets" / "detection_merged"
OUTPUT_ROOT = PROJECT_ROOT / "detection" / "outputs"
IMAGE_SIZE = 256
NUM_CLASSES = 3
CLASS_NAMES = ["cat", "dog", "panda"]
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
sys.path.insert(0, str(PROJECT_ROOT))
from detection.dataset import YOLODetectionDataset, detection_collate
from detection.model import EELANDetector
from detection.train_detection import build_targets, detection_loss
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
print("device:", DEVICE)
print("data:", DATA_ROOT)
```

```python
# Load and inspect the YOLO dataset.
train_ds = YOLODetectionDataset(DATA_ROOT, "train", IMAGE_SIZE)
valid_ds = YOLODetectionDataset(DATA_ROOT, "valid", IMAGE_SIZE)
test_ds = YOLODetectionDataset(DATA_ROOT, "test", IMAGE_SIZE)
print({"train": len(train_ds), "valid": len(valid_ds), "test": len(test_ds)})
sample_image, sample_target, sample_path = train_ds[0]
print("sample image:", tuple(sample_image.shape))
print("sample boxes:", sample_target.shape, sample_path)
```

```python
# Check the E-ELAN detector forward pass and loss.
loader = DataLoader(train_ds, batch_size=2, shuffle=False, collate_fn=detection_collate)
images, targets, paths = next(iter(loader))
model = EELANDetector(num_classes=NUM_CLASSES).to(DEVICE)
prediction = model(images.to(DEVICE))
encoded = build_targets(targets, prediction.shape[-1], NUM_CLASSES, DEVICE)
loss = detection_loss(prediction, encoded)
print("prediction:", tuple(prediction.shape))
print("initial loss:", float(loss))
```

```python
def run_epoch(model, loader, optimizer=None):
    training = optimizer is not None
    model.train(training)
    total = 0.0
    for images, targets, _ in loader:
        images = images.to(DEVICE)
        prediction = model(images)
        encoded = build_targets(targets, prediction.shape[-1], NUM_CLASSES, DEVICE)
        batch_loss = detection_loss(prediction, encoded)
        if training:
            optimizer.zero_grad(set_to_none=True)
            batch_loss.backward()
            optimizer.step()
        total += batch_loss.item() * images.size(0)
    return total / len(loader.dataset)

# Overfit test: the loss should decrease strongly on 10 images.
overfit_ds = Subset(train_ds, list(range(min(10, len(train_ds)))))
overfit_loader = DataLoader(overfit_ds, batch_size=2, shuffle=True, collate_fn=detection_collate)
overfit_model = EELANDetector(num_classes=NUM_CLASSES).to(DEVICE)
overfit_optimizer = torch.optim.AdamW(overfit_model.parameters(), lr=1e-3)
overfit_losses = []
for epoch in range(1, 101):
    value = run_epoch(overfit_model, overfit_loader, overfit_optimizer)
    overfit_losses.append(value)
    if epoch == 1 or epoch % 10 == 0:
        print(f"overfit epoch {epoch:03d}: loss={value:.4f}")
print("initial_loss=", overfit_losses[0], "final_loss=", overfit_losses[-1])

```

```python
# Full training. Run this only after the overfit loss decreases.
train_loader = DataLoader(train_ds, batch_size=16, shuffle=True, collate_fn=detection_collate)
valid_loader = DataLoader(valid_ds, batch_size=16, shuffle=False, collate_fn=detection_collate)
model = EELANDetector(num_classes=NUM_CLASSES).to(DEVICE)
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, weight_decay=1e-4)
history = []
best_valid = float("inf")
EPOCHS = 15
for epoch in range(1, EPOCHS + 1):
    train_loss = run_epoch(model, train_loader, optimizer)
    with torch.no_grad():
        valid_loss = run_epoch(model, valid_loader)
    row = {"epoch": epoch, "train_loss": train_loss, "valid_loss": valid_loss}
    history.append(row)
    print(f"epoch {epoch:03d}: train_loss={train_loss:.4f} valid_loss={valid_loss:.4f}")
    if valid_loss < best_valid:
        best_valid = valid_loss
        torch.save({"model": model.state_dict(), "classes": CLASS_NAMES, "history": history}, OUTPUT_ROOT / "best_detection.pt")
(OUTPUT_ROOT / "history.json").write_text(json.dumps(history, indent=2), encoding="utf-8")
print("saved:", OUTPUT_ROOT / "best_detection.pt")

```

```python
import matplotlib.pyplot as plt
from torchvision.ops import nms

def decode_predictions(output, confidence_threshold=0.25, iou_threshold=0.5, max_detections=100):
    batch_size, _, grid_h, grid_w = output.shape
    result = []
    raw = output.permute(0, 2, 3, 1)
    for batch_index in range(batch_size):
        item = raw[batch_index]
        yy, xx = torch.meshgrid(
            torch.arange(grid_h, device=output.device),
            torch.arange(grid_w, device=output.device), indexing="ij"
        )
        centers = torch.stack((xx, yy), dim=-1).float()
        centers = (centers + item[..., 0:2].clamp(0.0, 1.0)) / torch.tensor(
            [grid_w, grid_h], device=output.device
        )
        sizes = item[..., 2:4].clamp(0.0, 1.0)
        xyxy = torch.cat((centers - sizes / 2, centers + sizes / 2), dim=-1)
        objectness = item[..., 4].sigmoid()
        class_scores = item[..., 5:].sigmoid()
        class_confidence, labels = class_scores.max(dim=-1)
        scores = objectness * class_confidence
        keep = scores >= confidence_threshold
        boxes, scores, labels = xyxy[keep].clamp(0.0, 1.0), scores[keep], labels[keep]
        if boxes.numel():
            kept_parts = []
            for class_id in labels.unique().tolist():
                class_indices = torch.where(labels == class_id)[0]
                kept_parts.append(class_indices[nms(boxes[class_indices], scores[class_indices], iou_threshold)])
            kept = torch.cat(kept_parts) if kept_parts else torch.empty(0, dtype=torch.long)
            boxes, scores, labels = boxes[kept], scores[kept], labels[kept]
        if scores.numel() > max_detections:
            top = scores.argsort(descending=True)[:max_detections]
            boxes, scores, labels = boxes[top], scores[top], labels[top]
        result.append({"boxes": boxes.cpu(), "scores": scores.cpu(), "labels": labels.cpu()})
    return result

def box_iou_xyxy(boxes_a, boxes_b):
    if boxes_a.numel() == 0 or boxes_b.numel() == 0:
        return torch.zeros((len(boxes_a), len(boxes_b)))
    top_left = torch.maximum(boxes_a[:, None, :2], boxes_b[None, :, :2])
    bottom_right = torch.minimum(boxes_a[:, None, 2:], boxes_b[None, :, 2:])
    intersection = (bottom_right - top_left).clamp(min=0).prod(dim=-1)
    area_a = (boxes_a[:, 2:] - boxes_a[:, :2]).clamp(min=0).prod(dim=-1)
    area_b = (boxes_b[:, 2:] - boxes_b[:, :2]).clamp(min=0).prod(dim=-1)
    return intersection / (area_a[:, None] + area_b[None, :] - intersection + 1e-8)

def evaluate_detector(model, dataset, confidence_threshold=0.25, iou_threshold=0.5, max_detections=100):
    model.eval()
    true_positive = false_positive = false_negative = 0
    detections = {class_id: [] for class_id in range(NUM_CLASSES)}
    ground_truth_count = {class_id: 0 for class_id in range(NUM_CLASSES)}
    with torch.no_grad():
        for index in range(len(dataset)):
            image, target, _ = dataset[index]
            prediction = decode_predictions(
                model(image.unsqueeze(0).to(DEVICE)), confidence_threshold, iou_threshold, max_detections
            )[0]
            gt_centers, gt_sizes = target[:, 1:3], target[:, 3:5]
            gt_boxes = torch.cat((gt_centers - gt_sizes / 2, gt_centers + gt_sizes / 2), dim=1)
            gt_labels = target[:, 0].long()
            for class_id in range(NUM_CLASSES):
                ground_truth_count[class_id] += int((gt_labels == class_id).sum())
            matched = set()
            for box, score, label in zip(prediction["boxes"], prediction["scores"], prediction["labels"]):
                candidates = torch.where(gt_labels == label)[0]
                best_iou, best_index = 0.0, -1
                for candidate in candidates.tolist():
                    if candidate not in matched:
                        iou = float(box_iou_xyxy(box[None], gt_boxes[candidate][None])[0, 0])
                        if iou > best_iou:
                            best_iou, best_index = iou, candidate
                is_tp = best_iou >= iou_threshold and best_index >= 0
                detections[int(label)].append((float(score), is_tp))
                if is_tp:
                    matched.add(best_index)
                    true_positive += 1
                else:
                    false_positive += 1
            false_negative += len(gt_boxes) - len(matched)
    precision = true_positive / max(true_positive + false_positive, 1)
    recall = true_positive / max(true_positive + false_negative, 1)
    f1 = 2 * precision * recall / max(precision + recall, 1e-8)
    average_precisions = []
    for class_id in range(NUM_CLASSES):
        ranked = sorted(detections[class_id], key=lambda item: item[0], reverse=True)
        tp = fp = 0
        precisions, recalls = [], []
        for _, is_tp in ranked:
            tp += int(is_tp)
            fp += int(not is_tp)
            precisions.append(tp / max(tp + fp, 1))
            recalls.append(tp / max(ground_truth_count[class_id], 1))
        ap = sum(max((p for p, r in zip(precisions, recalls) if r >= threshold), default=0.0) for threshold in [step / 10 for step in range(11)]) / 11
        average_precisions.append(ap)
    return {
        "precision": precision, "recall": recall, "f1": f1,
        "mAP@50": sum(average_precisions) / NUM_CLASSES,
        "AP@50": dict(zip(CLASS_NAMES, average_precisions)),
        "counts": {"TP": true_positive, "FP": false_positive, "FN": false_negative},
    }






```

```python
# Load the checkpoint with the lowest validation loss.
checkpoint = torch.load(OUTPUT_ROOT / "best_detection.pt", map_location=DEVICE)
eval_model = EELANDetector(num_classes=NUM_CLASSES).to(DEVICE)
eval_model.load_state_dict(checkpoint["model"])
metrics = evaluate_detector(eval_model, test_ds, confidence_threshold=0.75, iou_threshold=0.5, max_detections=20)
print(f"Precision: {metrics['precision']:.4f}")
print(f"Recall:    {metrics['recall']:.4f}")
print(f"F1-score:  {metrics['f1']:.4f}")
print(f"mAP@50:    {metrics['mAP@50']:.4f}")
print("AP@50:", {name: round(value, 4) for name, value in metrics["AP@50"].items()})
print("Counts:", metrics["counts"])

```

```python
# Plot train/validation loss and visualize one test prediction.
with (OUTPUT_ROOT / "history.json").open(encoding="utf-8") as file:
    saved_history = json.load(file)
epochs = [row["epoch"] for row in saved_history]
plt.figure(figsize=(7, 4))
plt.plot(epochs, [row["train_loss"] for row in saved_history], label="train loss")
plt.plot(epochs, [row["valid_loss"] for row in saved_history], label="valid loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Detection training loss")
plt.legend()
plt.grid(alpha=0.25)
plt.show()

def show_prediction(dataset, index=0, confidence_threshold=0.25):
    image, target, path = dataset[index]
    with torch.no_grad():
        prediction = decode_predictions(
            eval_model(image.unsqueeze(0).to(DEVICE)), confidence_threshold, 0.5
        )[0]
    display_image = image.permute(1, 2, 0).numpy()
    display_image = display_image * [0.229, 0.224, 0.225] + [0.485, 0.456, 0.406]
    display_image = display_image.clip(0, 1)
    plt.figure(figsize=(7, 7))
    plt.imshow(display_image)
    for box, score, label in zip(prediction["boxes"], prediction["scores"], prediction["labels"]):
        x1, y1, x2, y2 = box.tolist()
        rect = plt.Rectangle((x1 * IMAGE_SIZE, y1 * IMAGE_SIZE), (x2 - x1) * IMAGE_SIZE, (y2 - y1) * IMAGE_SIZE, fill=False, color="red", linewidth=2)
        plt.gca().add_patch(rect)
        plt.text(x1 * IMAGE_SIZE, y1 * IMAGE_SIZE, f"{CLASS_NAMES[int(label)]} {float(score):.2f}", color="white", backgroundcolor="red")
    plt.axis("off")
    plt.title(Path(path).name)
    plt.show()

show_prediction(test_ds, index=0)
```

```python
# Find a practical confidence threshold on the validation set.
threshold_results = []
for threshold in [0.25, 0.35, 0.45, 0.55, 0.65, 0.75]:
    result = evaluate_detector(eval_model, valid_ds, confidence_threshold=threshold, iou_threshold=0.5)
    threshold_results.append((threshold, result))
    print(f"threshold={threshold:.2f} precision={result['precision']:.4f} recall={result['recall']:.4f} f1={result['f1']:.4f} mAP@50={result['mAP@50']:.4f}")
best_threshold, best_threshold_metrics = max(threshold_results, key=lambda item: item[1]["f1"])
print(f"selected threshold={best_threshold:.2f} by validation F1={best_threshold_metrics['f1']:.4f}")
```
